# Data Cleaning

```python
# Drop Row Id Column
df = df.drop('Row ID', axis=1)
display(df.head())
```

```python
# Replace any null values in postal code column to "N/A"

df['Postal Code'] = df['Postal Code'].fillna('N/A')
print(f"Null values in 'Postal Code' after filling: {df['Postal Code'].isnull().sum()}")
```

```python
# Check to see if any rows in the sales or quantinty columns have negative values
print(f"Negative values in 'Sales' column: {df[df['Sales'] < 0].shape[0]}")
print(f"Negative values in 'Quantity' column: {df[df['Quantity'] < 0].shape[0]}")
```

```python
# Show some of the rows that have negative quantity values
df[df['Quantity'] < 0].head(10)
```

```python
# Drop rows where 'Quantity' is negative
df = df[df['Quantity'] >= 0]
print(f"Shape of DataFrame after dropping rows with negative Quantity: {df.shape}")
```

```python
# Check for duplicate rows
duplicate_rows = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_rows}")

df = df.drop_duplicates()
print(f"Shape of DataFrame after dropping duplicates: {df.shape}")
```

## Inspecting Categorical Columns

```python
# Identify categorical columns
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(f"\n--- Column: {col} ---")
    print(f"Number of unique values: {df[col].nunique()}")
    display(df[col].value_counts())
```

### Cleaning 'Ship Mode' Column

```python
# Standardize 'Ship Mode' values
ship_mode_mapping = {
    'Standard  Class': 'Standard Class',
    '2nd Class': 'Second Class',
    'First class': 'First Class',
    'Same-Day': 'Same Day'
}

df['Ship Mode'] = df['Ship Mode'].replace(ship_mode_mapping)

print("Value counts for 'Ship Mode' after standardization:")
display(df['Ship Mode'].value_counts())
```

### Continuing Inspection of Categorical Columns

```python
# List of columns already inspected (or cleaned):
inspected_cols = ['Order ID', 'Ship Mode', 'Customer ID']

# Identify remaining categorical columns to inspect
all_categorical_cols = df.select_dtypes(include='object').columns
remaining_categorical_cols = [col for col in all_categorical_cols if col not in inspected_cols]

print("Inspecting remaining categorical columns:")
for col in remaining_categorical_cols:
    print(f"\n--- Column: {col} ---")
    print(f"Number of unique values: {df[col].nunique()}")
    display(df[col].value_counts())
```

### Cleaning 'Customer Name' Column

```python
# Standardize 'Customer Name' by converting to title case and stripping whitespace
df['Customer Name'] = df['Customer Name'].str.strip().str.title()

print("Value counts for 'Customer Name' after standardization:")
display(df['Customer Name'].value_counts())
```

### Continued Inspection of Remaining Categorical Columns

```python
# Update the list of columns already cleaned):
inspected_cols = ['Order ID', 'Ship Mode', 'Customer ID', 'Customer Name', 'Segment']

# Identify remaining categorical columns to inspect
all_categorical_cols = df.select_dtypes(include='object').columns
remaining_categorical_cols = [col for col in all_categorical_cols if col not in inspected_cols]

print("Inspecting remaining categorical columns:")
for col in remaining_categorical_cols:
    print(f"\n--- Column: {col} ---")
    print(f"Number of unique values: {df[col].nunique()}")
    display(df[col].value_counts())
```

### Cleaning 'City' and 'State/Province' Columns

```python
# Standardize 'City' and 'State/Province' by converting to title case and stripping whitespace
df['City'] = df['City'].str.strip().str.title()
df['State/Province'] = df['State/Province'].str.strip().str.title()

print("Value counts for 'City' after standardization (top 10):")
display(df['City'].value_counts().head(10))

print("\nValue counts for 'State/Province' after standardization (top 10):")
display(df['State/Province'].value_counts().head(10))
```

### Continuing Inspection of Remaining Categorical Columns

```python
# Update the list of columns already inspected (or cleaned):
inspected_cols = ['Order ID', 'Ship Mode', 'Customer ID', 'Customer Name', 'Segment', 'Country/Region', 'City', 'State/Province']

# Identify remaining categorical columns to inspect
all_categorical_cols = df.select_dtypes(include='object').columns
remaining_categorical_cols = [col for col in all_categorical_cols if col not in inspected_cols]

print("Inspecting remaining categorical columns:")
for col in remaining_categorical_cols:
    print(f"\n--- Column: {col} ---")
    print(f"Number of unique values: {df[col].nunique()}")
    display(df[col].value_counts())
```

### Final Inspection of Remaining Categorical Columns

```python
# Update the list of columns already inspected (or cleaned):
inspected_cols = ['Order ID', 'Ship Mode', 'Customer ID', 'Customer Name', 'Segment', 'Country/Region', 'City', 'State/Province', 'Postal Code', 'Region', 'Product ID']

# Identify remaining categorical columns to inspect
all_categorical_cols = df.select_dtypes(include='object').columns
remaining_categorical_cols = [col for col in all_categorical_cols if col not in inspected_cols]

print("Inspecting remaining categorical columns:")
for col in remaining_categorical_cols:
    print(f"\n--- Column: {col} ---")
    print(f"Number of unique values: {df[col].nunique()}")
    display(df[col].value_counts())
```

# Cleaning the "Returns" excel sheet

```python
df3 = pd.read_excel('superstore_expanded.xlsx',sheet_name = 'Returns')
df3.head(10)
```

```python
# Check for duplicates in df3
duplicate_count = df3.duplicated().sum()
print(f"Number of duplicate rows in df3: {duplicate_count}")

if duplicate_count > 0:
    df3 = df3.drop_duplicates()
    print(f"Duplicates removed. New shape: {df3.shape}")

# Convert 'Returned' column to boolean
# 'Yes' becomes True, anything else (like NaN or No) becomes False
df3['Returned'] = df3['Returned'].apply(lambda x: True if str(x).strip().lower() == 'yes' else False)

print("\n'Returned' column converted to boolean.")
display(df3.head())
display(df3.dtypes)
```
